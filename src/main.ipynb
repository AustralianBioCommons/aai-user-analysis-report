## Amount of Different Types of Email Address

How many types of email addresses have been used to sign up?

```python
different_emails = set()
for i in range(df["Email"].size):
    email = df["Email"].iloc[i].split("@", 1)[-1]
    different_emails.add(email)

len(different_emails)
```

## Mean, Mode, and Range Amounts of Logins

Mean

```python
df["Logins Count"] = df["Logins Count"].fillna(0)
column_size = df["Logins Count"].size

total_logins = 0
for i in range(column_size):
    total_logins += df["Logins Count"].iloc[i]

average = total_logins / column_size
print(f"The mean amount of logins is {average.round(3)}")
```

Mode

```python
login_times = {}
for i in range(df["Logins Count"].size):
    val = df["Logins Count"].iloc[i]
    login_times[val] = login_times.get(val, 0) + 1

sorted_hashmap = dict(sorted(login_times.items()))

first_key, first_val = next(iter(sorted_hashmap.items()))

print(f"The mode amount of logins is {first_key")
```
