## Amount of Different Types of Email Address

How many types of email addresses have been used to sign up?

```python
different_emails = set()
for i in range(df["Email"].size):
    email = df["Email"].iloc[i].split("@", 1)[-1]
    different_emails.add(email)

len(different_emails)
```
